#### Psicologo

```python
# Lista
psicologos = []

# Agregar
def agregarpsicologo():

    nombre = input("Nombre: ")
    telefono = input("Teléfono: ")
    edad = input("Edad: ")

    psicologos.append([nombre, telefono, edad])

    print("Agregado")


# Leer
def leerpsicologo():

    nombre = input("Nombre: ")

    for p in psicologos:

        if p[0] == nombre:

            print(p)

            return

    print("No encontrado")


# Listar
def listarpsicologos():

    for p in psicologos:

        print(p)


# Buscar
def buscarpsicologo():

    nombre = input("Buscar: ")

    for p in psicologos:

        if p[0] == nombre:

            print("Encontrado", p)

            return

    print("No existe")


# Modificar
def modificarpsicologo():

    nombre = input("Modificar: ")

    for p in psicologos:

        if p[0] == nombre:

            p[0] = input("Nuevo nombre: ")
            p[1] = input("Nuevo teléfono: ")
            p[2] = input("Nueva edad: ")

            print("Modificado")

            return

    print("No encontrado")


# Borrar
def borrarpsicologo():

    nombre = input("Borrar: ")

    for p in psicologos:

        if p[0] == nombre:

            psicologos.remove(p)

            print("Eliminado")

            return

    print("No encontrado")


# Menú
while True:

    print("\n1.Agregar")
    print("2.Leer")
    print("3.Listar")
    print("4.Buscar")
    print("5.Modificar")
    print("6.Borrar")
    print("7.Salir")

    op = input("Opción: ")

    if op == "1":
        agregarpsicologo()

    elif op == "2":
        leerpsicologo()

    elif op == "3":
        listarpsicologos()

    elif op == "4":
        buscarpsicologo()

    elif op == "5":
        modificarpsicologo()

    elif op == "6":
        borrarpsicologo()

    elif op == "7":

        print("Programa realizado por Hernandez Dylan 1245 NL = 24 2-H")

        break

    else:
        print("Error")
```
